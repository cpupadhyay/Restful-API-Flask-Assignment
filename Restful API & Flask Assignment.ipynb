# Theoritical Questions

1. What is a RESTful API?
   -> A RESTful API (Representational State Transfer) is a web service that uses HTTP methods (GET, POST, PUT, DELETE) to access and manipulate resources        identified by URLs.  
      It follows stateless communication and typically returns data in formats like JSON or XML.

2. Explain the concept of API specification.
   -> An API specification is a formal document that defines how an API behaves, including its endpoints, request/response formats, data types,                  authentication methods, and error handling.It serves as a contract between the API provider and consumer, ensuring consistent and predictable              integration.

3. What is Flask, and why is it popular for building APIs?
   -> Flask is a lightweight and flexible Python web framework used to build web applications and APIs.  
      It's popular for API development because of its simplicity, minimal setup, and extensive support through extensions, making it easy to create           RESTful services quickly.

4. What is routing in Flask?
   -> Routing in Flask refers to the process of mapping URLs to specific functions in your application.  
      Each route is defined using a Python decorator (`@app.route()`), which tells Flask what code to execute when a particular URL is accessed.

```python
5. How do you create a simple Flask application?
   -> Creating a simple Flask application involves these basic steps:
      -> Import Flask – Bring in the Flask class from the `flask` module.  
      -> Create an application instance – Instantiate the Flask app, usually with `Flask(__name__)`.  
      -> Define routes – Use decorators like `@app.route()` to map URLs to Python functions (views) that handle requests and return responses.  
      -> Run the application – Use `app.run()` to start the development server and make the app accessible in a web browser.
      This structure allows Flask to handle incoming web requests and return appropriate responses based on defined routes.
```

```python
6. What are HTTP methods used in RESTful APIs?
   -> The main HTTP methods used in RESTful APIs are:
      ->GET – Retrieve data from the server.  
      ->POST – Send new data to the server to create a resource.  
      ->PUT – Update an existing resource entirely.  
      ->PATCH – Partially update an existing resource.  
      ->DELETE – Remove a resource from the server.
```

7. What is the purpose of the @app.route() decorator in Flask?
   -> The `@app.route()` decorator in Flask is used to bind a specific URL path to a Python function, defining what should happen when that URL is            accessed.It tells Flask which function to execute in response to a request to that route.

8. What is the difference between GET and POST HTTP methods?
   -> The GET method is used to retrieve data from a server without affecting it, and parameters are sent in the URL.The POST method is used to send          data to the server to create or process resources, with data included in the request body, making it more secure for sensitive information.

9. How do you handle errors in Flask APIs?
   -> Error handling in Flask APIs involves the following approaches:
      -> Try-Except Blocks: Errors are often caught using Python's `try-except` mechanism. This allows you to handle specific exceptions within your             routes and return custom error messages or status codes in the response.
      -> Custom Error Handlers: Flask provides a way to globally handle errors (like 404 for "Not Found" or 500 for "Internal Server Error") by using            the `@app.errorhandler()` decorator. This enables a consistent way to respond to various error scenarios across the entire application.
      -> Abort Function: The `abort()` function is a Flask utility to raise specific HTTP errors (like 404 or 403). When invoked, it immediately                 terminates the request and returns the specified error code to the client.

10. How do you connect Flask to a SQL database?
    -> Connecting Flask to a SQL database involves the following steps:
       ->Install a Database Library: First, you need a Python library to interact with the database, such as `SQLAlchemy` (for ORM-based interaction)            or `psycopg2` (for direct PostgreSQL access). For SQLAlchemy, you typically install the `Flask-SQLAlchemy` extension, which simplifies                  integration.
       ->Configure Database URI: Set the connection string (or URI) that Flask will use to connect to the database. This typically includes the                  username, password, host, and database name.

         python code:
         app.config['SQLALCHEMY_DATABASE_URI'] = 'sqlite:///mydatabase.db'  # Example for SQLite

       ->Initialize the Database Object: If using SQLAlchemy, initialize the database extension by creating an instance of `SQLAlchemy` and binding it           to the Flask app.

         python code:
         from flask_sqlalchemy import SQLAlchemy
         db = SQLAlchemy(app)

       ->Define Models: Define the database tables as Python classes (models) using SQLAlchemy. Each class corresponds to a table in the database, and           each attribute corresponds to a column.

         python code:
         class User(db.Model):
         id = db.Column(db.Integer, primary_key=True)
         username = db.Column(db.String(80), unique=True, nullable=False)
         email = db.Column(db.String(120), unique=True, nullable=False)

       ->Create the Database: After defining your models, you would create the database by running migrations or using SQLAlchemy’s `create_all()`               method.

         python code:
         with app.app_context():
         db.create_all()

       ->Perform Database Operations: Use the ORM (Object-Relational Mapping) features of SQLAlchemy to interact with the database (e.g., adding,                querying, updating, or deleting records).

         python code:
         new_user = User(username='example', email='example@example.com')
         db.session.add(new_user)
         db.session.commit()

11. What is the role of Flask-SQLAlchemy?
    -> Flask-SQLAlchemy is an extension that integrates SQLAlchemy with Flask, providing an easy way to work with databases using ORM. It simplifies           database configuration, session management, and query handling within Flask applications.

12. What are Flask blueprints, and how are they useful?

13. What is the purpose of Flask's request object?
    -> Flask's `request` object is used to access data sent with an incoming HTTP request, such as form data, query parameters, JSON payload, headers,         and more.It allows you to interact with the request content and extract necessary information for processing the request in your Flask                  application.

14. How do you create a RESTful API endpoint using Flask?

15. What is the purpose of Flask's jsonify() function?
    -> Flask's `jsonify()` function is used to convert Python dictionaries or lists into JSON format, making it easy to return JSON responses from             Flask routes. It also sets the correct `Content-Type` header (`application/json`) for the response.

16. Explain Flask’s url_for() function.
    -> In theory, Flask's `url_for()` function is used to dynamically generate URLs for a given view function by its endpoint name. Instead of                 hardcoding URLs, `url_for()` ensures that the correct URL is generated based on the route definitions and can include variables or query                parameters. This makes your application more maintainable and less prone to errors when URLs change.

17. How does Flask handle static files (CSS, JavaScript, etc.)?
    -> Flask serves static files (like CSS and JavaScript) from a default `static` folder, accessible via the `/static/` URL path. You can link to             these files in templates using `url_for('static', filename='...')`.

18. What is an API specification, and how does it help in building a Flask API?
    -> An API specification is a detailed document that defines the endpoints, methods, request/response formats, parameters, and error codes for an           API. It helps in building a Flask API by providing a clear blueprint for both developers and consumers to understand how to interact with the           API consistently and predictably.

19. What are HTTP status codes, and why are they important in a Flask API?
    -> HTTP status codes are three-digit numbers returned by the server to indicate the outcome of an HTTP request. They are divided into categories           like success (2xx), client errors (4xx), and server errors (5xx).In a Flask API, status codes are important because they help clients understand        the result of their request (e.g., success, failure, invalid input) and guide appropriate actions or error handling based on the response code.

20. How do you handle POST requests in Flask?
    -> Handling POST requests in Flask involves defining a route that specifies the `POST` method in the `@app.route()` decorator. The data                    sent with the request, typically in the body (such as JSON or form data), is accessed using Flask's `request` object. The handler processes this        data and returns a response, often in JSON format, with an appropriate HTTP status code to indicate the result of the operation.

21. How would you secure a Flask API?
    -> To secure a Flask API, implement authentication (e.g., JWT, OAuth) and authorization, use HTTPS, and validate inputs to prevent attacks.                Additionally, apply rate limiting, configure CORS, handle errors securely, and monitor logs for suspicious activity.

22. What is the significance of the Flask-RESTful extension?
    -> The Flask-RESTful extension simplifies the creation of RESTful APIs in Flask by providing tools for defining resources, handling HTTP                   methods (GET, POST, PUT, DELETE), and automatically converting responses to JSON. It streamlines routing, input parsing, and error handling,            making it easier to build and maintain APIs in Flask.

23. What is the role of Flask’s session object?
    -> Flask's `session` object is used to store data across requests for a specific user, enabling you to maintain state between different pages.
```python
# 2. How do you serve static files like images or CSS in Flask?
'''
In Flask, static files like images, CSS, or JavaScript are typically served from a special folder named `static`. Flask automatically serves files from this folder using the `/static/` URL path.

-> Steps to Serve Static Files:

->Create a `static` folder: Inside your project directory, create a folder named `static`. Place your static files (e.g., `style.css`, `image.jpg`) in this folder.

->Link static files in your templates: Use Flask's `url_for()` function to link to static files dynamically in your HTML templates.'''

# ->Example:

# Directory structure:
/your_project
    /static
        /images
            logo.jpg
        /css
            style.css
    /templates
        index.html
    app.py


# app.py:
# python:
from flask import Flask, render_template

app = Flask(__name__)

@app.route('/')
def index():
    return render_template('index.html')

if __name__ == '__main__':
    app.run(debug=True)


# index.html:
# html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Flask Static Files</title>
    <link rel="stylesheet" href="{{ url_for('static', filename='css/style.css') }}">
</head>
<body>
    <h1>Welcome to Flask</h1>
    <img src="{{ url_for('static', filename='images/logo.jpg') }}" alt="Logo">
</body>
</html>
```

```python
# 3. How do you define different routes with different HTTP methods in Flask?
# code
from flask import Flask, request, jsonify

app = Flask(__name__)

# GET method: Retrieve data
@app.route('/resource', methods=['GET'])
def get_resource():
    return jsonify({"message": "GET request to fetch resource"})

# POST method: Create data
@app.route('/resource', methods=['POST'])
def create_resource():
    data = request.get_json()
    return jsonify({"message": "POST request to create resource", "data": data}), 201

# PUT method: Update data
@app.route('/resource', methods=['PUT'])
def update_resource():
    data = request.get_json()
    return jsonify({"message": "PUT request to update resource", "data": data})

# DELETE method: Delete data
@app.route('/resource', methods=['DELETE'])
def delete_resource():
    return jsonify({"message": "DELETE request to remove resource"})

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 6. How do you handle forms in Flask?
#Code
# app.py
# python
from flask import Flask, request, render_template

app = Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method == 'POST':
        name = request.form['name']
        return f'Hello, {name}!'
    return render_template('form.html')

if __name__ == '__main__':
    app.run(debug=True)

# templates/form.html
'''html
<!DOCTYPE html>
<html>
<head>
    <title>Simple Form</title>
</head>
<body>
    <form method="POST" action="/submit">
        <label for="name">Enter your name:</label>
        <input type="text" name="name" required>
        <input type="submit" value="Submit">
    </form>
</body>
</html>
'''
```

```python
# 7. How can you validate form data in Flask?
# Code
# Manual Validation
# Use simple condition checks in your route
# python
from flask import Flask, request, render_template

app = Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    error = None
    if request.method == 'POST':
        name = request.form.get('name')
        if not name:
            error = "Name is required."
        else:
            return f'Hello, {name}!'
    return render_template('form.html', error=error)

# Validation with Flask-WTF
''' Install Flask-WTF:
bash
pip install flask-wtf '''


# app.py
# python
from flask import Flask, render_template
from flask_wtf import FlaskForm
from wtforms import StringField, SubmitField
from wtforms.validators import DataRequired

app = Flask(__name__)
app.secret_key = 'secret_key'  # Needed for CSRF protection

class NameForm(FlaskForm):
    name = StringField('Name', validators=[DataRequired()])
    submit = SubmitField('Submit')

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    form = NameForm()
    if form.validate_on_submit():
        return f'Hello, {form.name.data}!'
    return render_template('form.html', form=form)

# templates/form.html
'''html
<form method="POST">
    {{ form.hidden_tag() }}
    {{ form.name.label }} {{ form.name(size=20) }}
    {% if form.name.errors %}
        <p style="color:red;">{{ form.name.errors[0] }}</p>
    {% endif %}
    {{ form.submit() }}
</form>'''
```

```python
# 8. How do you manage sessions in Flask?
# Code
# How to Use Sessions in Flask:
# Set a secret key (required to sign session cookies):
# python
app = Flask(__name__)
app.secret_key = 'your_secret_key'

# Store data in a session:
# python
from flask import session

@app.route('/login')
def login():
    session['username'] = 'john_doe'
    return 'User logged in!'

# Access session data:
# python
@app.route('/profile')
def profile():
    username = session.get('username')
    return f'Hello, {username}' if username else 'Not logged in.'


# Remove data from session (logout):
# python
@app.route('/logout')
def logout():
    session.pop('username', None)
    return 'User logged out.'
```

```python
# 11. How do you structure a Flask app using Blueprints?
# Code
# Basic Structure:

/myapp
    /auth
        __init__.py
        routes.py
    /main
        __init__.py
        routes.py
    app.py
# step-by-step:
-> auth/routes.py:
# python
from flask import Blueprint

auth = Blueprint('auth', __name__)

@auth.route('/login')
def login():
    return "Login Page"

-> main/routes.py:
# python
from flask import Blueprint

main = Blueprint('main', __name__)

@main.route('/')
def home():
    return "Home Page"

-> app.py:
# python
from flask import Flask
from auth.routes import auth
from main.routes import main

app = Flask(__name__)

# Register Blueprints
app.register_blueprint(auth, url_prefix='/auth')
app.register_blueprint(main)

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 12. How do you define a custom Jinja filter in Flask?
# Code
# Step-by-Step Example:

->Define the filter using a decorator:
# python
from flask import Flask

app = Flask(__name__)

# Define a custom filter to reverse a string
@app.template_filter('reverse')
def reverse_filter(s):
    return s[::-1]

->Use it in a Jinja template:

# templates/index.html
'''html
<!DOCTYPE html>
<html>
<body>
    <p>Original: {{ name }}</p>
    <p>Reversed: {{ name|reverse }}</p>
</body>
</html>'''

->Render the template:
# python
from flask import render_template

@app.route('/')
def home():
    return render_template('index.html', name='Flask')
```
